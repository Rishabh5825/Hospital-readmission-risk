```python
import requests
import json
```

```python
# The URL of your running FastAPI server
url = "http://127.0.0.1:8000/predict"
```

```python
headers = {
    "X-API-Key": "hospital-secure-key-2026"
}
```

```python
# A fake patient profile (Notice: No race or gender!)
patient_data = {
    "age": 75,
    "admission_type_id": "Emergency",
    "discharge_disposition_id": "Discharged_Home",
    "admission_source_id": "Emergency_Room",
    "time_in_hospital": 5,
    "num_lab_procedures": 44,
    "num_procedures": 1,
    "num_medications": 13,
    "number_outpatient": 0,
    "number_emergency": 0,
    "number_inpatient": 2,
    "number_diagnoses": 9,
    "diag_1": "Circulatory",
    "diag_2": "Diabetes",
    "diag_3": "Respiratory",
    "max_glu_serum": "Not_Measured",
    "A1Cresult": "Not_Measured",
    "insulin": "Up",
    "change": "Ch",
    "diabetesMed": "Yes"
}
```

```python
# Send the POST request
print("Sending patient data to API...")
response = requests.post(url, json=patient_data, headers=headers)

# Print the response beautifully
if response.status_code == 200:
    print("\n✅ API Response:")
    print(json.dumps(response.json(), indent=4))
else:
    print(f"❌ Error {response.status_code}: {response.text}")
```
